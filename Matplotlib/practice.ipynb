```python
import matplotlib.pyplot as plt

# Create simple x and y values
x = [0, 1, 2, 3, 4, 5]
y = x  # Since y = x

# Plot the line
plt.plot(x, y, label='y = x')

# Add labels and title
plt.title('Line Plot: y = x')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,10])
y = np.exp(x)
plt.plot(x,y,label  ='e^x',color = 'blue',linestyle= '-.',linesize = "10")
plt.legend()

plt.xlabel('X')
plt.ylabel("Y = e^x")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 10]
y = [i**2 for i in x]  # Square each element in x

plt.plot(x, y, label='x^2')
plt.legend()

plt.xlabel('X')
plt.ylabel('Y = X^2')

plt.title('Plot of Y = X^2')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

countries = ['USA', 'India', 'Brazil', 'Germany', 'Japan']
population = [331, 1388, 213, 83, 126]

# Bar chart with hatch patterns
plt.bar(countries, population, hatch='/*+OX')  # Apply a repeating pattern

# Add text labels on top of bars
for i in range(len(population)):
    plt.text(i, population[i], population[i], ha='center', va='bottom')

plt.ylabel('Population (millions)')
plt.show()
```
