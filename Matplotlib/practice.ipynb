```python
import matplotlib.pyplot as plt

# Create simple x and y values
x = [0, 1, 2, 3, 4, 5]
y = x  # Since y = x

# Plot the line
plt.plot(x, y, label='y = x')

# Add labels and title
plt.title('Line Plot: y = x')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,10])
y = np.exp(x)
plt.plot(x,y,label  ='e^x',color = 'blue',linestyle= '-.',linesize = "10")
plt.legend()

plt.xlabel('X')
plt.ylabel("Y = e^x")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 10]
y = [i**2 for i in x]  # Square each element in x

plt.plot(x, y, label='x^2')
plt.legend()

plt.xlabel('X')
plt.ylabel('Y = X^2')

plt.title('Plot of Y = X^2')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
categories = ['A', 'B', 'C', 'D']
values = [100, 15, 7, 12]

# Plot
plt.bar(categories, values)

plt.legend()

# Add labels and title
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Simple Bar Chart')

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate x values from 0 to 2π
x = np.linspace(0, 2 * np.pi, 100)

# Compute y = sin(x)
y = np.sin(x)

# Plot the graph
plt.plot(x, y)

# Add labels and title
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Graph of sin(x)')

# Add a grid
plt.grid(True)

# Show the plot
plt.show()
```
